```python
import numpy as np

# Create an array
array = np.array([83, 25, 19, 47, 62, 15, 96, 39, 51])

print("Original Array:")
print(array)

# Square root of array elements
print("\nSquare Root:")
print(np.sqrt(array))

# Check dimensions
print("\nDimension of Array:")
print(array.ndim)

# Reshape array into 3x3
new_array = array.reshape(3, 3)

print("\nReshaped Array:")
print(new_array)

# Check dimensions after reshaping
print("\nDimension after Reshaping:")
print(new_array.ndim)

# Convert 2D array back to 1D
print("\nRavelled Array:")
print(new_array.ravel())

# Array slicing
print("\nSelected Elements:")
print(new_array[2, 1:3])
```
