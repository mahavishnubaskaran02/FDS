```python
import numpy as np
import matplotlib.pyplot as plt

# Sample dataset
data = np.array([10, 12, 15, 14, 13, 18, 20, 22, 19, 50])

# Calculate Q1 and Q3
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Detect outliers
outliers = data[(data < lower_limit) | (data > upper_limit)]

print("Dataset:")
print(data)

print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("\nOutliers:")
print(outliers)

# Box plot
plt.boxplot(data)
plt.title("Outlier Detection using Box Plot")
plt.ylabel("Values")
plt.grid(False)
plt.show()
```
