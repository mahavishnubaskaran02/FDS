```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales_2025 = [20, 30, 25, 40, 45]
sales_2026 = [25, 35, 30, 45, 50]

plt.plot(months, sales_2025, marker="o", label="2025")
plt.plot(months, sales_2026, marker="o", label="2026")

plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Comparison")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch", "Headphones"]
sales = [70, 45, 30, 55, 40]

plt.bar(products, sales)
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Product Sales - Bar Chart")

plt.show()
```

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch", "Headphones"]
sales = [35, 25, 15, 15, 10]

plt.pie(sales, labels=products, autopct='%1.1f%%')
plt.title("Product Sales - Pie Chart")

plt.show()
```
